````python
#算法复杂度 :O(nlogn)
# 更高效的归并排序

def mergeSort(arr):
    # 如果数组长度大于 1，则继续进行分割排序
    if len(arr) > 1:
        middle = len(arr) // 2
        left = arr[:middle]
        right = arr[middle:]

        mergeSort(left)
        mergeSort(right)

        left_idx = right_idx = arr_idx = 0

        # 合并左右两个已排序的子数组
        while left_idx < len(left) and right_idx < len(right):
            if left[left_idx] < right[right_idx]:
                arr[arr_idx] = left[left_idx]
                left_idx += 1
            else:
                arr[arr_idx] = right[right_idx]
                right_idx += 1
            arr_idx += 1

        while left_idx < len(left):
            arr[arr_idx] = left[left_idx]
            left_idx += 1
            arr_idx += 1

        while right_idx < len(right):
            arr[arr_idx] = right[right_idx]
            right_idx += 1
            arr_idx += 1

test_arr = [10, 7, 11, 2, 5, 6]
mergeSort(test_arr)
print("归并排序结果:", test_arr)

````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:通过一次遍历来找到最大利润。首先初始化最小买入价格为第一天的价格，最大利润为0。

遍历数组，对于每一天：如果当前价格小于已知的最小买入价格，更新最小买入价格。
否则，计算当前价格与最小买入价格的差值，如果大于已知的最大利润，则更新最大利润和对应的买入卖出日期。

这种方法的时间复杂度是O(n)。

````python

# 算法复杂度:O(n)
def max_profit(profits):
    if not profits or len(profits) < 2:
        return 0, None, None  # 无法交易的情况

    min_buy = profits[0]  # 初始化最小买入价格
    max_profit_value = 0  # 初始化最大利润
    buy_day = 0  # 初始化买入日期（0-based）
    sell_day = 1  # 初始化卖出日期（0-based）
    temp_buy_day = 0  # 临时记录当前最小买入价格的日期

    for i in range(1, len(profits)):
        if profits[i] < min_buy:
            min_buy = profits[i]  # 更新最小买入价格
            temp_buy_day = i  # 更新临时买入日期
        else:
            current_profit = profits[i] - min_buy
            if current_profit > max_profit_value:
                max_profit_value = current_profit  # 更新最大利润
                buy_day = temp_buy_day  # 更新买入日期
                sell_day = i  # 更新卖出日期

    if max_profit_value <= 0:
        return 0, None, None  # 没有正利润的交易
    else:
        return max_profit_value, buy_day + 1, sell_day + 1
# example
profits1 = [3,2,1,-7,5,2,-1,3,-1]
profit_value1, buy1, sell1 = max_profit(profits1)
print(f"最大利润: {profit_value1}, 买入日期: 第{buy1}天, 卖出日期: 第{sell1}天")

````

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:1、设计思路：可以采用二分查找法来解决这个问题。

第一步初始化：
左指针 left = 1（假设索引从 1 开始）。
右指针 right = n-1（A[n] 无右邻居）。

第二步计算中间位置 mid。检查 A[mid] 是否满足局部最小条件：

如果 A[mid] < A[mid-1] 且 A[mid] < A[mid+1]，直接返回 A[mid]。
否则：如果 A[mid] > A[mid-1]，说明局部最小值在左半边（left 到 mid-1）。

如果 A[mid] > A[mid+1]，说明局部最小值在右半边（mid+1 到 right）。

第三步终止条件：当 left == right 时，A[left] 即为局部最小值

````python

#算法复杂度:O(nlogn)

# 二分查找法
def _min_(A):
    n = len(A)
    left, right = 1, n - 2  # 忽略首尾
    while left <= right:
        mid = (left + right) // 2
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]  # 找到局部最小值
        elif A[mid] > A[mid - 1]:
            right = mid - 1  # 向左搜索
        else:
            left = mid + 1  # 向右搜索
    return A[left]  # 边界情况（left == right）

# example
A = [9, 3, 7, 2, 1, 4, 5 ]
print(_min_(A))
````

idea:1. 排序后取前k个数

设计思路：排序数组：使用高效的排序算法（快速排序）对数组进行排序。
取前k个数：排序后直接输出前 k 个元素。
排序时间复杂度：O(nlogn)（快速排序）。

总时
2.使用堆（Heap）提取前k个数
设计思路：a.构建最小堆：将数组转换为最小堆（堆顶是最小元素）。使用heapify 方法（从最后一个非叶子节点开始，自底向上调整堆。）
b.提取前k个数：执行 k 次 extract-min 操作（即每次取出堆顶元素，然后重新调整堆）。
c.输出结果：按提取顺序返回 k 个最小元素（因为每次取出的都是当前堆中的最小值）。


````python
# add your code here
# 时间复杂度:O(n+klogn)
# 第一问，采用了快排
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)

def _smallest_quick(arr, k):
    arr_sorted = quicksort(arr)
    return arr_sorted[:k]

# example
arr = [5,4,3,2,6,1,88,33,22,107]
print(_smallest_quick(arr, 3))
# your algorithm time complexity is :O(logn)
````

````python
# 使用堆（Heap）提取前k个数
import heapq

def _smallest_heap(arr, k):
    heapq.heapify(arr)  # 原地构建最小堆（O(n)）
    return [heapq.heappop(arr) for _ in range(k)]  # 提取前k个最小元素（O(k log n)）

# example
arr = [5,4,3,2,6,1,88,33,22,107]
print(_smallest_heap(arr.copy(), 3))  # 输出: [1, 2, 3]
# your algorithm time complexity is :O(n+klogn)
````

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

---

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:使用一个栈来模拟星球的碰撞过程。

遍历数组，对于每个星球：当栈为空或者当前星球向右运动（值为正）时，将当前星球入栈。
当栈不为空且当前星球向左运动（值为负）时，比较栈顶星球和当前星球：
若栈顶星球质量小于当前星球质量，弹出栈顶元素，继续与新的栈顶元素比较，直到栈为空或者栈顶星球质量大于当前星球质量。
若栈顶星球质量等于当前星球质量，弹出栈顶元素，当前星球也消失（不做入栈操作）。
若栈顶星球质量大于当前星球质量，当前星球消失（不做入栈操作）。

````python

# your algorithm time complexity is:O(n)
def planet_collision(planets):
    surviving_planets = []  # 替代原来的stack

    for current_planet in planets:
        # 处理向右运动的星球
        if current_planet > 0:
            surviving_planets.append(current_planet)
            continue

        # 处理向左运动的星球及其可能的碰撞
        collision_occurred = False
        while surviving_planets and surviving_planets[-1] > 0:
            right_moving_planet = surviving_planets[-1]

            if right_moving_planet < -current_planet:
                surviving_planets.pop()  # 右侧星球被摧毁
            elif right_moving_planet == -current_planet:
                surviving_planets.pop()  # 两者质量相同，都摧毁
                collision_occurred = True
                break
            else:
                collision_occurred = True  # 当前星球被摧毁
                break

        if not collision_occurred:
            surviving_planets.append(current_planet)

    return surviving_planets

A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
A2 = [23, -8, 9, -3, -7, 9, -23, 22]

print(planet_collision(A1))
print(planet_collision(A2))
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:0 - 1 背包问题要在固定容量背包里选物品使总价值最大，且物品只能选或不选。用动态规划把大问题拆成小问题求解，避免重复计算。

对于每一个物品，我们都有两种选择。

不放入物品：如果不把这个物品放入背包，那么当前能获得的最大价值就和不考虑这个物品时一样。
放入物品：前提是背包还有足够的容量来装这个物品。放入后，能获得的最大价值等于不考虑这个物品且背包剩余容量时的最大价值加上这个物品的价值。我们会在这两种情况中选择能让价值更大的那个作为当前状态下的最大价值。

时间复杂度：O(nW)，n是物品数，W是背包容量。

````python
# add your code here
# your algorithm time complexity is:O(nW)

def maximize_value(weights, profits, max_capacity):
    n = len(weights)
    # 创建一个一维数组 dp 来存储子问题的解
    dp = [0] * (max_capacity + 1)

    # 填充 dp 数组
    for i in range(1, n + 1):
        for w in range(max_capacity, 0, -1):
            if weights[i - 1] <= w:
                # 选择放入或不放入当前物品，取价值最大的情况
                dp[w] = max(profits[i - 1] + dp[w - weights[i - 1]], dp[w])
            # 当前物品重量超过当前背包容量，不能放入，直接跳过

    return dp[max_capacity]


# 物品的重量列表
weights = [5, 4, 6, 3]
# 物品的价值列表
profits = [10, 40, 30, 50]
# 背包的容量
max_capacity = 9

# 调用函数求解最大价值
max_value = maximize_value(weights, profits, max_capacity)
print("背包能装下的最大价值为:", max_value)
````

4、递归或迭代：可以使用递归或迭代的方法来实现中序遍历和节点值的交换。

该算法的时间复杂度为 O(n)，其中 n 是树中节点的数量。这是因为我们需要遍历整棵树来找到被交换的节点。

````python
# add your code here
# your algorithm time complexity is:O(n)
# 恢复二叉搜索树（deepseek）
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def recoverTree(self, root: TreeNode) -> None:
        """
        Do not return anything, modify root in-place instead.
        """
        # Morris中序遍历实现
        first = second = None
        prev = TreeNode(float('-inf'))

        current = root
        while current:
            if not current.left:
                # 检查当前节点
                if prev.val > current.val:
                    if not first:
                        first = prev
                    second = current
                prev = current
                current = current.right
            else:
                # 找到前驱节点
                predecessor = current.left
                while predecessor.right and predecessor.right != current:
                    predecessor = predecessor.right

                if not predecessor.right:
                    predecessor.right = current
                    current = current.left
                else:
                    predecessor.right = None
                    # 检查当前节点
                    if prev.val > current.val:
                        if not first:
                            first = prev
                        second = current
                    prev = current
                    current = current.right

        # 交换两个错误节点的值
        first.val, second.val = second.val, first.val

# 辅助函数：构建树
def build_tree(nodes):
    if not nodes:
        return None
    root = TreeNode(nodes[0])
    queue = [root]
    i = 1
    while queue and i < len(nodes):
        current = queue.pop(0)
        if nodes[i] is not None:
            current.left = TreeNode(nodes[i])
            queue.append(current.left)
        i += 1
        if i < len(nodes) and nodes[i] is not None:
            current.right = TreeNode(nodes[i])
            queue.append(current.right)
        i += 1
    return root

# 辅助函数：中序遍历验证
def inorder_traversal(root):
    result = []
    stack = []
    current = root
    while current or stack:
        while current:
            stack.append(current)
            current = current.left
        current = stack.pop()
        result.append(current.val)
        current = current.right
    return result

# 测试用例1
nodes1 = [1, 3, None, None, 2]
root1 = build_tree(nodes1)
print("原始树中序遍历:", inorder_traversal(root1))
Solution().recoverTree(root1)
print("恢复后中序遍历:", inorder_traversal(root1))

# 测试用例2
nodes2 = [3, 1, 4, None, None, 2]
root2 = build_tree(nodes2)
print("\n原始树中序遍历:", inorder_traversal(root2))
Solution().recoverTree(root2)
print("恢复后中序遍历:", inorder_traversal(root2))
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

<div style="text-align:center;width:800px;"> <img alt="" src="./fig/hw3q6.png"> </div>

请分析算法设计思路，计算时间复杂度，并基于python编程实现

算法设计思路:可以通过动态规划来解决。算法的基本思路是从两边向中间遍历，计算每个位置能够接住的雨水量。
1、初始化：
创建两个数组 left 和 right 来分别存储每个位置左侧和右侧的最大高度。
创建一个数组 water 来存储每个位置可以接的雨水量。
2、计算左侧最大高度：
从左到右遍历 heights 数组，更新 left 数组，使得 left[i] 存储到第 i 个位置为止左侧的最大高度。
计算右侧最大高度：
3、从右到左遍历 heights 数组，更新 right 数组，使得 right[i] 存储到第 i 个位置为止右侧的最大高度。
计算雨水量：
4、遍历 heights 数组，对于每个位置 i，计算该位置可以接的雨水量为 min(left[i], right[i]) - heights[i]。
如果计算结果为负，则该位置不能接雨水，water[i] 保持为0。
5、计算总雨水量：
遍历 water 数组，将所有位置的雨水量相加，得到总雨水量。
该算法的时间复杂度为 O(n)，其中 n 是输入数组的长度。这是因为我们只需要遍历一次数组。

````python
# add your code here
# your algorithm time complexity is:O(n)
 接雨水问题
def trap_rainwater(heights):
    if not heights or len(heights) < 3:
        return 0

    n = len(heights)
    left = [0] * n
    right = [0] * n
    water = [0] * n

    # 初始化左侧最大高度数组
    left[0] = heights[0]
    for i in range(1, n):
        left[i] = max(left[i - 1], heights[i])

    # 初始化右侧最大高度数组
    right[n - 1] = heights[n - 1]
    for i in range(n - 2, -1, -1):
        right[i] = max(right[i + 1], heights[i])

    # 计算每个位置的雨水量
    for i in range(1, n - 1):
        water[i] = min(left[i], right[i]) - heights[i]
        water[i] = max(water[i], 0)

    # 计算总雨水量
    return sum(water)

# 示例 1
heights1 = [0,1,0,2,1,0,1,3,2,1,2,1]
print("example 1 output:", trap_rainwater(heights1))

# 示例 2
heights2 = [4,2,0,3,2,5]
print("example 2 output:", trap_rainwater(heights2))
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:

 输入：root = [1,2,3], 输出：25

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:这个问题可以通过深度优先搜索(DFS)来解决，因为我们需要遍历所有从根到叶节点的路径。
1、从根节点开始进行DFS遍历；2、在遍历过程中维护当前路径表示的数字；3、当到达叶节点时，将当前数字加到总和中；4、遍历完成后返回总和
（定义Node类，含节点值及左右子节点引用；
使用build_tree函数按层序遍历列表构建树，用队列辅助，依次添加左右子节点；
最后计算和：get_sum函数含递归函数walk。walk函数遇空节点返回 0；更新当前路径和；遇叶节点返回当前和；否则递归计算左右子树路径和并相加。）
时间复杂度分析：需要访问每个节点一次：O(n)，其中n是树中的节点数；对于每个节点，数字操作（乘以10加当前值）是O(1)操作
总时间复杂度：O(n)

````python
# add your code here
# 算法复杂度:O(n）
# 从根到叶节点数字之和
class Node:
    def __init__(self, num=0, lefty=None, righty=None):
        self.num = num
        self.lefty = lefty
        self.righty = righty


def get_sum(tree):
    def walk(n, cur):
        if not n:
            return 0
        cur = cur * 10 + n.num
        if not n.lefty and not n.righty:
            return cur
        return walk(n.lefty, cur) + walk(n.righty, cur)

    return walk(tree, 0)


def build_tree(arr):
    if not arr:
        return None
    top = Node(arr[0])
    que = [top]
    idx = 1
    while que and idx < len(arr):
        now = que.pop(0)
        if arr[idx] is not None:
            now.lefty = Node(arr[idx])
            que.append(now.lefty)
        idx = idx + 1
        if idx < len(arr) and arr[idx] is not None:
            now.righty = Node(arr[idx])
            que.append(now.righty)
        idx = idx + 1
    return top


# example 1
data1 = [1, 2, 3]
tree1 = build_tree(data1)
print("Result 1:", get_sum(tree1))

# example 2
data2 = [4, 9, 0, 5, 1]
tree2 = build_tree(data2)
print("Result 2:", get_sum(tree2))

````
